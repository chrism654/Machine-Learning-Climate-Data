```python
import pandas as pd
import numpy as np
import seaborn as sns
import os
import operator
import matplotlib.pyplot as plt
import matplotlib.ticker as ticker
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
from matplotlib.pyplot import figure
```

```python
#Create a path to data
path = r'/Users/chrismace/Desktop/7 Machine Learning'
```

```python
df = pd.read_csv(os.path.join(path, 'Dataset-weather-prediction-dataset-processed.csv'))
```

```python
df.head()
```

```python
df.shape
```

```python
#Reduce your dataset to a single year
df1984 = df[df['DATE'].astype(str).str.contains('1984')] #<-----INSERT YEAR HERE
df1984
```

```python
#Scale the weather features
features = df1984
scaled_features = StandardScaler().fit_transform(features)
```

```python
scaled_features
```

```python
#Run the PCA model, reducing to 18 components
pca = PCA(n_components=18)
principalComponents = pca.fit_transform(scaled_features)
```

```python
principalComponents
```

```python
#Make a dataframe of the new components
dfPCA = pd.DataFrame(principalComponents, columns=['PCA1', 'PCA2', 'PCA3', 'PCA4', 'PCA5', 'PCA6', 'PCA7', 'PCA8', 'PCA9', 'PCA10',
                                                   'PCA11', 'PCA12', 'PCA13', 'PCA14', 'PCA15', 'PCA16','PCA17', 'PCA18'])
```

```python
dfPCA
```

```python
dfPCA.to_csv(os.path.join(path, 'PCA_1984.csv'))
```
